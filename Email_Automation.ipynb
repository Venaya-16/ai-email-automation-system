```python
from google.colab import userdata
import gradio as gr
```

```python
import google.generativeai as genai
genai.configure(api_key=userdata.get('GOOGLE_API_KEY'))
model = genai.GenerativeModel('gemini-3.5-flash')
```

```python
def ask_gemini(prompt):
    response = model.generate_content(prompt)
    return response.text
```

```python
def generate_email(email_type, recipient, purpose, tone):

    prompt = f"""
You are an expert business communication assistant.

Write a professional email.

Email Type:
{email_type}

Recipient:
{recipient}

Purpose:
{purpose}

Tone:
{tone}

Requirements:

- Add an appropriate subject line.
- Begin with a professional greeting.
- Write a clear and concise email.
- Maintain the requested tone.
- End with a suitable closing.
"""

    return ask_gemini(prompt)
```

```python
def reply_email(email, tone):

    prompt = f"""
Read the following email and write a professional reply.

Tone:
{tone}

Email:

{email}

Requirements:

- Keep the context.
- Be polite.
- Maintain professionalism.
- Return only the reply email.
"""

    return ask_gemini(prompt)
```

```python
def rewrite_email(email):

    prompt = f"""
Improve the following email.

Fix:

- Grammar
- Professionalism
- Readability
- Formatting

Return only the improved email.

Email:

{email}
"""

    return ask_gemini(prompt)
```

```python
def change_tone(email, tone):

    prompt = f"""
Rewrite the following email.

Keep the meaning the same.

Change the tone to:

{tone}

Return only the rewritten email.

Email:

{email}
"""

    return ask_gemini(prompt)
```

```python
theme = gr.themes.Soft(
    primary_hue="green",
    secondary_hue="emerald"
)

with gr.Blocks(
    title="📧 AI Email Automation System"
) as demo:

    gr.Markdown(
        """
# 📧 AI Email Automation System

Generate professional emails, automate replies, improve drafts and personalize tone using Google Gemini.
"""
    )

    with gr.Accordion("✉️ Generate New Email", open=True):

        email_type = gr.Dropdown(
            ["HR","Sales","Support","General"],
            label="Email Type"
        )

        recipient = gr.Textbox(
            label="Recipient Name"
        )

        purpose = gr.Textbox(
            label="Purpose",
            lines=3
        )

        tone = gr.Dropdown(
            ["Professional","Friendly","Formal","Persuasive"],
            label="Tone"
        )

        generate_btn = gr.Button(
            "Generate Email",
            variant="primary"
        )

        generated_email = gr.Markdown()

        generate_btn.click(
            generate_email,
            inputs=[
                email_type,
                recipient,
                purpose,
                tone
            ],
            outputs=generated_email
        )

    with gr.Accordion("💬 Reply to Email"):

        incoming_email = gr.Textbox(
            label="Paste Email",
            lines=10
        )

        reply_tone = gr.Dropdown(
            ["Professional","Friendly","Formal"],
            label="Reply Tone"
        )

        reply_btn = gr.Button("Generate Reply")

        reply_output = gr.Markdown()

        reply_btn.click(
            reply_email,
            inputs=[
                incoming_email,
                reply_tone
            ],
            outputs=reply_output
        )

    with gr.Accordion("📝 Rewrite Email"):

        draft = gr.Textbox(
            label="Paste Draft",
            lines=10
        )

        rewrite_btn = gr.Button("Rewrite Email")

        rewritten = gr.Markdown()

        rewrite_btn.click(
            rewrite_email,
            inputs=draft,
            outputs=rewritten
        )

    with gr.Accordion("🎭 Change Tone"):

        email = gr.Textbox(
            label="Paste Email",
            lines=10
        )

        new_tone = gr.Dropdown(
            [
                "Professional",
                "Friendly",
                "Formal",
                "Persuasive",
                "Apologetic"
            ],
            label="New Tone"
        )

        tone_btn = gr.Button("Convert Tone")

        tone_output = gr.Markdown()

        tone_btn.click(
            change_tone,
            inputs=[
                email,
                new_tone
            ],
            outputs=tone_output
        )

    gr.Markdown(
       \
    )
```

```python
demo.launch(theme=theme)
```
